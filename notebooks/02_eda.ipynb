```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

injury = pd.read_csv("../data/raw/InjuryRecord.csv")
playlist = pd.read_csv("../data/raw/PlayList.csv")
tracking = pd.read_csv("../data/raw/PlayerTrackData.csv")

print("All datasets loaded successfully!")
```

```python
print(injury.head())

print("\nInjury distribution:")
print(injury["BodyPart"].value_counts())

print("\nInjury records:")
print(injury.shape[0])
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=injury,
    x="BodyPart",
    hue="BodyPart",
    legend=False
)

plt.title("Distribution of Injuries by Body Part")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print(playlist["FieldType"].value_counts())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=playlist,
    x="FieldType",
    hue="FieldType",
    legend=False
)

plt.title("Distribution of Playing Surfaces")
plt.show()
```

```python
print(tracking.columns.tolist())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    tracking["s"],
    bins=50,
    kde=True
)

plt.title("Player Speed Distribution")
plt.xlabel("Speed")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    tracking["o"],
    bins=50,
    kde=True
)

plt.title("Player Acceleration Distribution")
plt.xlabel("Acceleration")
plt.show()
```
