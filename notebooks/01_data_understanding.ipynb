```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
injury = pd.read_csv("../data/raw/InjuryRecord.csv")
playlist = pd.read_csv("../data/raw/PlayList.csv")
tracking = pd.read_csv("../data/raw/PlayerTrackData.csv")
```

```python
print("Injury Dataset:", injury.shape)
print("PlayList Dataset:", playlist.shape)
print("Tracking Dataset:", tracking.shape)
```

```python
for name, df in {
    "Injury": injury,
    "PlayList": playlist,
    "Tracking": tracking
}.items():

    print("\n", "=" * 50)
    print(name)
    print("=" * 50)

    print("Columns:", df.columns.tolist())
    print("\nData Types:")
    print(df.dtypes)
    print("\nFirst 5 Rows:")
    display(df.head())
```

```python
for name, df in {
    "Injury": injury,
    "PlayList": playlist,
    "Tracking": tracking
}.items():

    print(f"\n{name}")
    print(df.isnull().sum().sort_values(ascending=False))
```

```python
print("Injury duplicates:", injury.duplicated().sum())
print("Playlist duplicates:", playlist.duplicated().sum())
print("Tracking duplicates:", tracking.duplicated().sum())
```
